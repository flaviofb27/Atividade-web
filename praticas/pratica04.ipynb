## Um exemplo do que entregar

**Pergunta:** quantos brasileiros usam a internet, e quantos ainda estão fora? **Público:** uma secretaria que planeja inclusão digital. **Fonte:** Our World in Data (ourworldindata.org/grapher/share-of-individuals-using-the-internet).

**Gráfico:** uma linha de 2000 a 2024, com o Brasil em cor e o mundo em cinza. Título: *"Em 24 anos, o Brasil foi de 3% para 84% de pessoas conectadas"*.

**A história, em três falas (uma por integrante):**

- **Começo** (o contexto): *"Uma secretaria quer planejar a inclusão digital. Fomos ver quantos brasileiros usam a internet, com dados do Our World in Data."*
- **Meio** (o achado): *"Em 2000, só 3 em cada 100 brasileiros usavam a internet. Em 2024, são 84. O Brasil passou a média do mundo, mas ainda está atrás de Chile, Uruguai e Argentina."*
- **Fim** (o "e daí?"): *"Faltam 16 em cada 100 brasileiros. O próximo passo é descobrir quem são e onde estão, porque é para eles que a política precisa olhar."*

## Passo 6 · O gráfico e a história
Façam **um** gráfico. Para comparar categorias, **barras**; para mostrar mudança no tempo, **linha**. Ele precisa de:

- **um destaque só**: o ponto do achado em cor, o resto em cinza;
- **título que diz a conclusão**, não o assunto ("O Brasil foi de 3% para 84%", e não "Uso da internet");
- **barras começando no zero**;
- **fonte** no rodapé;
- nada que não ajude a mensagem (grade, legenda sobrando, cores sem significado).

Depois, escrevam numa célula de texto **a história do trio**, com duas ou três frases por fala e o nome de quem fala cada parte:

- **Começo:** o contexto (a pergunta, para quem, de onde vêm os dados);
- **Meio:** o achado, com os números;
- **Fim:** o "e daí?" (o que o público deve pensar ou fazer, e o que os dados não mostram).

```python
import matplotlib.pyplot as plt

# 1. Criando a figura do gráfico
plt.figure(figsize=(11, 6))

# 2. Definindo cores
cores = {
    'Brazil': '#00897B',        # Verde destacado para o Brasil
    'United States': '#002868', # Azul para os EUA
    'China': '#de2910',         # Vermelho para a China
    'World': '#757575'          # Cinza para a Média Mundial
}

nomes_pt = {
    'Brazil': 'Brasil',
    'United States': 'EUA',
    'China': 'China',
    'World': 'Média Mundial'
}

# 3. Plotando as linhas de cada país/entidade
for pais, cor in cores.items():
    dados_pais = df_filtrado[df_filtrado['entity'] == pais]

    # Plota a linha (destacando o Brasil com uma linha mais grossa)
    plt.plot(dados_pais['year'], dados_pais['emissions_total_per_capita'],
             color=cor, linewidth=3.0 if pais == 'Brazil' else 1.8)

    # Adiciona o texto com o nome no final de cada linha
    ultimo = dados_pais.iloc[-1]
    plt.text(ultimo['year'] + 0.4, ultimo['emissions_total_per_capita'],
             f"{nomes_pt[pais]} ({ultimo['emissions_total_per_capita']:.1f}t)",
             color=cor, fontsize=9, va='center', fontweight='bold' if pais == 'Brazil' else 'normal')

# 4. Título focado no achado (conclusão)
plt.title("Em 2025, cada brasileiro emitiu menos da metade de CO₂ da média mundial",
          fontsize=13, fontweight='bold', loc='left', pad=15)

# 5. Configurações dos eixos e limites
plt.ylabel("Toneladas de CO₂ por pessoa", fontsize=10, color='#424242')
plt.xlim(1990, 2030)
plt.ylim(0, 22)

# 6. Limpando o design (removendo bordas desnecessárias)
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#BDBDBD')
ax.spines['bottom'].set_color('#BDBDBD')
plt.grid(axis='y', linestyle='--', alpha=0.3)

# 7. Rodapé com a fonte dos dados
plt.figtext(0.125, 0.01, "Fonte: Our World in Data / Global Carbon Budget (1990–2025). Nota: Emissões por combustíveis fósseis e indústria.",
            fontsize=8, color='#757575')

# Ajuste final de layout e exibição
plt.tight_layout(rect=[0, 0.04, 1, 1])
plt.show()
```

**Em 2025, cada brasileiro emitiu 2,3 toneladas de CO2 por combustíveis fósseis e indústria — menos da metade da média mundial, que é de 4,7 toneladas, e muito abaixo da China (8,8t) e dos Estados Unidos (14,5t). Além disso, o Brasil atingiu seu pico em 2014, com 2,78 toneladas, e conseguiu recuar.**

```python
import plotly.express as px

# Identificar o ano mais recente na base de dados
ano_recente = df['year'].max()

# Filtrar os dados para o ano mais recente e remover agrupamentos regionais
df_mapa = df[(df['year'] == ano_recente) & (df['code'].notna()) & (df['code'] != '') & (df['code'].str.len() == 3)]

# Criar o mapa-múndi coroplético interativo
fig_mapa = px.choropleth(
    df_mapa,
    locations="code",
    color="emissions_total_per_capita",
    hover_name="entity",
    color_continuous_scale=px.colors.sequential.Reds,
    title=f"Emissões Globais de CO₂ per Capita por País ({ano_recente})",
    labels={'emissions_total_per_capita': 'CO₂ per capita (toneladas)'}
)

# Ajustes de layout para melhor visualização
fig_mapa.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type='equirectangular'
    ),
    margin={"r":0,"t":40,"l":0,"b":0}
)

fig_mapa.show()
```

**Esses números mostram que nossa matriz energética limpa coloca o Brasil em vantagem per capita, mas há um ponto cego importante: esta base não inclui o desmatamento.
